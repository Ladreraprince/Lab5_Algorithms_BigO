```python
import time, random

# Student Input Setup
LAST_NAME = input("LAST_NAME: ").strip()
STUDENT_ID = input("STUDENT_ID: ").strip()
FAVORITE_ARTIST = input("FAVORITE_ARTIST: ").strip()

ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LEN = len(LAST_NAME)
ARTIST_LEN = len(FAVORITE_ARTIST)

# Dataset Setup
base = [(i * NAME_LEN + ID_SUM) for i in range(1, 9)]
inputs = {
    "Sorted": sorted(base),
    "Reverse": sorted(base, reverse=True),
    "Random": random.sample(base, len(base))
}

def bubble_sort(arr):
    a, cmp_cnt, mov_cnt = arr[:], 0, 0
    t0 = time.perf_counter()
    n = len(a)
    for i in range(n):
        for j in range(0, n - i - 1):
            cmp_cnt += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                mov_cnt += 3
    return a, cmp_cnt, mov_cnt, time.perf_counter() - t0

def selection_sort(arr):
    a, cmp_cnt, mov_cnt = arr[:], 0, 0
    t0 = time.perf_counter()
    n = len(a)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            cmp_cnt += 1
            if a[j] < a[min_idx]: min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            mov_cnt += 3
    return a, cmp_cnt, mov_cnt, time.perf_counter() - t0

def insertion_sort(arr):
    a, cmp_cnt, mov_cnt = arr[:], 0, 0
    t0 = time.perf_counter()
    for i in range(1, len(a)):
        key = a[i]
        mov_cnt += 1
        j = i - 1
        while j >= 0:
            cmp_cnt += 1
            if a[j] > key:
                a[j + 1] = a[j]
                mov_cnt += 1
                j -= 1
            else: break
        a[j + 1] = key
        mov_cnt += 1
    return a, cmp_cnt, mov_cnt, time.perf_counter() - t0

# Execution & Results Output
print("\n--- Assessment Data ---")
print(f"Generated order dataset (Base): {base}\n")

for cond_name, data in inputs.items():
    print(f"=== Condition: {cond_name} {data} ===")
    for name, algo in [("Bubble Sort", bubble_sort), ("Selection Sort", selection_sort), ("Insertion Sort", insertion_sort)]:
        res, cmps, mvmt, exec_t = algo(data)
        verified = res == sorted(data)
        print(f"[{name}] Sorted: {verified} | Comparisons: {cmps} | Movements: {mvmt} | Time: {exec_t:.6f}s")
    print()

print("Big O Analysis:")
print("- Bubble / Insertion: Best O(n), Avg/Worst O(n^2)")
print("- Selection Sort: Best/Avg/Worst O(n^2)")
```
